```python
date['date']=pd.to_datetime(date['date'])
date.info()
```

```python
#Extract Year
date['date_year']=date['date'].dt.year

#here, dt is an accessor object that allows us to access the datetime properties of the 'date' column. We can use dt.year to extract the year from the datetime object and create a new column 'date_year' in the dataframe.
date.head()
```

```python
date['date_month_no']=date['date'].dt.month
date['date_month_name']=date['date'].dt.month_name()
date.head()
```

```python
date['date_day']=date['date'].dt.day
date.head()
```

```python
date['date_dayOfTheWeek']=date['date'].dt.dayofweek
date['date_dayOfTheWeek_name']=date['date'].dt.day_name()
date.head()
```

```python
#is Weekend?

date['date_is_weekend']=np.where(date['date_dayOfTheWeek_name'].isin(['Saturday','Sunday']),1,0)
date.head()
```

```python
date['date_week']=date['date'].dt.isocalendar().week
date.head()
```

```python
date['quarter'] = date['date'].dt.quarter
date['semester'] = np.where(date['quarter'].isin([1,2]), 1, 2)

date.head()
```

```python
#elapsed time bw dates
import datetime
today = datetime.datetime.today()
today
```

```python
today - date['date']
```

```python
(today - date['date']).dt.days
```

```python
# Months passed
months = (
    (today.year - date['date'].dt.year) * 12 +
    (today.month - date['date'].dt.month)
)

print(months)
```
