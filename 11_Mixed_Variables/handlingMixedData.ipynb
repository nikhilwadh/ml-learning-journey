```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv('titanic.csv')
```

```python
df.head()
```

```python
df['number'].unique()
```

```python
fig=df['number'].value_counts().plot.bar()
fig.set_title('Number of passengers travelling with')
```

```python
#extract numerical part
df['number_numerical']=pd.to_numeric(df['number'],errors='coerce', downcast='integer')

#errors='coerce' will convert non-numeric values to NaN.
#downcast integer will convert the float values to integer if possible.
```

```python
#extract categorical part
df['number_categorical']=np.where(df['number_numerical'].isnull(),df['number'],np.nan)

#the args in where function are: condition, value if true, value if false. In this case, if the number_numerical is null, then we will take the value from number column, else we will assign NaN.
```

```python
df['cabin_num']=df['Cabin'].str.extract('(\d+)') #Capturing the digits from the Cabin column using regex and creating a new column cabin_num.
df['cabin_cat']=df['Cabin'].str[0] #Capturing the first letter
```

```python
df.head()
```

```python
# extract the last bit of ticket as number
df['ticket_num'] = df['Ticket'].apply(lambda s: s.split()[-1])
df['ticket_num'] = pd.to_numeric(df['ticket_num'],
                                   errors='coerce',
                                   downcast='integer')

# extract the first part of ticket as category
df['ticket_cat'] = df['Ticket'].apply(lambda s: s.split()[0])
df['ticket_cat'] = np.where(df['ticket_cat'].str.isdigit(), np.nan,
                              df['ticket_cat'])

df.head(20)

#here s.split()[-1] will split the string in the Ticket column by space and take the last part of the string. This will be converted to numeric using pd.to_numeric() function. The first part of the string will be extracted using s.split()[0] and if it is a digit, it will be replaced with NaN, else it will be kept as a category.
```
