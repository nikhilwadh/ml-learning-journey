```python
import pandas as pd
import seaborn as sns
```

```python
df=pd.read_csv('train.csv')
df.head()
```

```python
#Catagorical Data
```

```python
#Countplot : Frequency Count
df['Survived'].value_counts().plot(kind='bar')
```

```python
#Pie chart
df['Pclass'].value_counts().plot(kind='pie')
```

```python
df['Sex'].value_counts().plot(kind='pie',autopct='%2f')
```

```python
#Numerical Data
```

```python
import matplotlib.pyplot as plt
```

```python
#Histogram : tells exact count of each value in the dataset
plt.hist(df['Age'])
```

```python
#Displot :tells probablity using Probability Density Function (PDF)
sns.distplot(df['Age'])
```

```python
#Boxplot : tells the outliers in the dataset
sns.boxplot(df['Age'])
```

```python
#Other functions on Numerical Data
```

```python
df['Age'].min()
```

```python
df['Age'].max()
```

```python
df['Age'].mean()
```

```python
df['Age'].skew() #Tells skewness of the data. If skewness is 0, then it is a normal distribution. If skewness is negative, then it is left skewed and if skewness is positive, then it is right skewed.
```
